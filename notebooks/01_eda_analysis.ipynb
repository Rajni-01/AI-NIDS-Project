```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Monday (Normal Traffic) CSV file load 
df_monday = pd.read_csv('../data/Monday-WorkingHours.pcap_ISCX.csv')

# Removing leading/trailing spaces from Column names
df_monday.columns = df_monday.columns.str.strip()

print("Monday Dataset Shape:", df_monday.shape)
df_monday.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Load Friday files safely right here for the plot
df_f_morn = pd.read_csv(r'../data/Friday-WorkingHours-Morning.pcap_ISCX.csv')
df_f_port = pd.read_csv(r'../data/Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv')
df_f_ddos = pd.read_csv(r'../data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv')

# Clean column names for all
for df in [df_f_morn, df_f_port, df_f_ddos]:
    df.columns = df.columns.str.strip()

# Combine all Friday labels
friday_labels = pd.concat([df_f_morn['Label'], df_f_port['Label'], df_f_ddos['Label']])

# Generate countplot
ax = sns.countplot(
    y=friday_labels, 
    order=friday_labels.value_counts().index, 
    palette="viridis",
    hue=friday_labels,
    legend=False
)

plt.title('Distribution of Friday Traffic (CIC-IDS2017)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Packets / Flows', fontsize=12)
plt.ylabel('Traffic Label', fontsize=12)

# Annotate values on bars
for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(
            f'{int(width):,}',
            xy=(width, p.get_y() + p.get_height() / 2),
            xytext=(5, 0),
            textcoords="offset points",
            ha='left', va='center',
            fontsize=10
        )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# List of valid dataframes actually loaded in memory
dataframes_list = []

# Collect all daily dataframes safely
for name in ['df_monday', 'df_tuesday', 'df_wednesday', 'df_thursday_morning', 'df_thursday_afternoon', 'df_f_morn', 'df_f_port', 'df_f_ddos']:
    if name in globals():
        dataframes_list.append(globals()[name])

# Concatenate into master dataset
if dataframes_list:
    master_df = pd.concat(dataframes_list, ignore_index=True)
    print("Master Dataset Created Successfully!")
    print("Master Dataset Shape:", master_df.shape)
    print("\nGlobal Class Distribution (Traffic Labels):")
    print(master_df['Label'].value_counts())
else:
    print("Please ensure your dataframes are loaded.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 7))

# Get value counts for ordering
label_counts = master_df['Label'].value_counts()

# Generate a horizontal countplot for the master dataset
ax = sns.countplot(
    y=master_df['Label'], 
    order=label_counts.index, 
    palette="magma",
    hue=master_df['Label'],
    legend=False
)

plt.title('Global Traffic Distribution Across Entire CIC-IDS2017 Dataset', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Flow Samples / Packets', fontsize=12)
plt.ylabel('Traffic Label', fontsize=12)

# Annotate count values on the bars for clarity
for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(
            f'{int(width):,}',
            xy=(width, p.get_y() + p.get_height() / 2),
            xytext=(5, 0),
            textcoords="offset points",
            ha='left', va='center',
            fontsize=9
        )

plt.tight_layout()
plt.show()
```
