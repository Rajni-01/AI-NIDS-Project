```python
import os
import pandas as pd
import numpy as np

folder = "/content/drive/MyDrive/AI_NIDS_Dataset/MachineLearningCVE"

output_folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Cleaned_Files"
os.makedirs(output_folder, exist_ok=True)

csv_files = sorted([
    f for f in os.listdir(folder)
    if f.lower().endswith(".csv")
])

for file in csv_files:
    input_path = os.path.join(folder, file)
    output_path = os.path.join(output_folder, file)

    df = pd.read_csv(input_path, encoding="utf-8-sig")

    #  clean Column names
    df.columns = df.columns.str.strip()

    # convert  Infinity to  NaN
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    # remove only affected rows
    df.dropna(
        subset=["Flow Bytes/s", "Flow Packets/s"],
        inplace=True
    )

    # save Cleaned file
    df.to_csv(output_path, index=False)

    print(file, " | Remaining rows:", len(df))

print("\nAll cleaned files saved successfully!")
```

```python
import os
import pandas as pd

input_folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Cleaned_Files"

output_folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Duplicates_Removed"
os.makedirs(output_folder, exist_ok=True)

csv_files = sorted([
    f for f in os.listdir(input_folder)
    if f.lower().endswith(".csv")
])

for file in csv_files:
    input_path = os.path.join(input_folder, file)
    output_path = os.path.join(output_folder, file)

    df = pd.read_csv(input_path)

    original_rows = len(df)

    # Remove duplicate rows
    df.drop_duplicates(keep="first", inplace=True)

    remaining_rows = len(df)
    removed_rows = original_rows - remaining_rows

    # Save the processed file
    df.to_csv(output_path, index=False)

    print("\nFile:", file)
    print("Original rows:", original_rows)
    print("Duplicates removed:", removed_rows)
    print("Remaining rows:", remaining_rows)

    del df

print("\nAll duplicate-free files saved successfully!")
```

```python
import os
import pandas as pd

folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Duplicates_Removed"

csv_files = sorted([
    f for f in os.listdir(folder)
    if f.lower().endswith(".csv")
])

reference_columns = None

for file in csv_files:
    path = os.path.join(folder, file)

    columns = pd.read_csv(
        path,
        nrows=0,
        encoding="utf-8-sig"
    ).columns.tolist()

    print("\nFile:", file)
    print("Total columns:", len(columns))

    if reference_columns is None:
        reference_columns = columns
        print("Reference file selected.")
    elif columns == reference_columns:
        print("Columns match: Yes")
    else:
        print("Columns match: No")

        missing_columns = set(reference_columns) - set(columns)
        extra_columns = set(columns) - set(reference_columns)

        print("Missing columns:", missing_columns)
        print("Extra columns:", extra_columns)

print("\nColumn verification completed!")
```

```python
import os
import pandas as pd

input_folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Duplicates_Removed"

output_folder = "/content/drive/MyDrive/AI_NIDS_Dataset/Final_Output"
os.makedirs(output_folder, exist_ok=True)

output_path = os.path.join(output_folder, "AI_NIDS_Cleaned_Merged.csv")

csv_files = sorted([
    f for f in os.listdir(input_folder)
    if f.lower().endswith(".csv")
])

first_file = True
total_rows = 0

for file in csv_files:
    input_path = os.path.join(input_folder, file)

    print("Merging:", file)

    for chunk in pd.read_csv(
        input_path,
        chunksize=100000,
        encoding="utf-8-sig"
    ):
        chunk.to_csv(
            output_path,
            mode="w" if first_file else "a",
            header=first_file,
            index=False
        )

        total_rows += len(chunk)
        first_file = False

    print("Completed:", file)

print("\nAll files merged successfully!")
print("Total merged rows:", total_rows)
print("Total columns: 79")
print("Saved file:", output_path)
```
